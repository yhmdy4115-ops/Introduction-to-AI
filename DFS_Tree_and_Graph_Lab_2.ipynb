```python
# ===== Example 1: DFS on a tree =====
tree = {
    "A": ["B", "C", "D"],
    "B": ["E", "F"],
    "C": [],
    "D": ["G"],
    "E": [],
    "F": ["H"],
    "G": [],
    "H": [],
}

ROOT, TREE_GOAL = "A", "H"
```

```python
from queue import LifoQueue

def dfs_tree(tree, root, goal):
    stack = LifoQueue()                      # built-in stack
    stack.put([root])                        # start with the path that holds only the root
    visited = []

    while not stack.empty():                 # while the stack is not empty
        path = stack.get()                   # take the newest path (LIFO)
        node = path[-1]                      # the node we are standing on
        visited.append(node)

        if node == goal:                     # reached the goal
            return path

        for child in reversed(tree[node]):   # reversed so the leftmost child comes out first
            stack.put(path + [child])        # new path = old path plus the child

    return None                              # no path to the goal exists
```

```python
tree_path = dfs_tree(tree, ROOT, TREE_GOAL)
print("Path from", ROOT, "to", TREE_GOAL, ":", tree_path)
```

```python
# ===== Example 2: DFS on a graph =====
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B", "G"],
    "E": ["B", "F"],
    "F": ["C", "E"],
    "G": ["D"],
}

START, GRAPH_GOAL = "A", "F"
```

```python
def dfs_graph(graph, start, goal):
    stack = [[start]]                        # a list used as a stack
    visited = []

    while stack:                             # while the stack is not empty
        path = stack.pop()                   # take the newest path (LIFO)
        node = path[-1]

        if node in visited:                  # already handled through another route
            continue                         # skip it

        visited.append(node)                 # mark it as visited

        if node == goal:                     # reached the goal
            return path

        for neighbor in reversed(graph[node]):   # reversed so the leftmost neighbor comes out first
            if neighbor not in visited:          # only go to neighbors we have not visited
                stack.append(path + [neighbor])

    return None
```

```python
graph_path = dfs_graph(graph, START, GRAPH_GOAL)
print("Path from", START, "to", GRAPH_GOAL, ":", graph_path)
```
