```python
# ===== Example 1: BFS on a tree =====
tree = {
    "A": ["B", "C", "D"],
    "B": ["E", "F"],
    "C": [],
    "D": ["G"],
    "E": [],
    "F": ["H"],
    "G": [],
    "H": [],
}

ROOT, TREE_GOAL = "A", "H"
```

```python
from collections import deque

def bfs_tree(tree, root, goal):
    queue = deque()                          # built-in double-ended queue
    queue.append([root])                     # start with the path that holds only the root
    visited = []

    while queue:                             # while the queue is not empty
        path = queue.popleft()               # take the oldest path (FIFO)
        node = path[-1]                      # the node we are standing on
        visited.append(node)

        if node == goal:                     # reached the goal
            return path

        for child in tree[node]:
            queue.append(path + [child])     # new path = old path plus the child

    return None                              # no path to the goal exists
```

```python
tree_path = bfs_tree(tree, ROOT, TREE_GOAL)
print("Path from", ROOT, "to", TREE_GOAL, ":", tree_path)
```

```python
# ===== Example 2: BFS on a graph =====
graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B", "G"],
    "E": ["B", "F"],
    "F": ["C", "E"],
    "G": ["D"],
}

START, GRAPH_GOAL = "A", "F"
```

```python
def bfs_graph(graph, start, goal):
    queue = [[start]]                        # a list used as a queue
    visited = []

    while queue:                             # while the queue is not empty
        path = queue.pop(0)                  # take the oldest path (FIFO)
        node = path[-1]

        if node in visited:                  # already handled through another route
            continue                         # skip it

        visited.append(node)                 # mark it as visited

        if node == goal:                     # reached the goal
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:      # only go to neighbors we have not visited
                queue.append(path + [neighbor])

    return None
```

```python
graph_path = bfs_graph(graph, START, GRAPH_GOAL)
print("Path from", START, "to", GRAPH_GOAL, ":", graph_path)
```
